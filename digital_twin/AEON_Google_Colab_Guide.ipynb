# 🌐 Passo 3: Ativar Servidor com Ngrok

Agora vamos configurar e iniciar o servidor AEON com acesso público através do **ngrok**.

## 🔧 Como funciona:
1. **Criamos um servidor Flask** que serve a interface AEON
2. **Ngrok cria um túnel público** para o servidor local
3. **Obtemos uma URL pública** para acessar o sistema
4. **Testamos as APIs** através de requisições HTTP

## ⚠️ Importante:
- O servidor ficará rodando em **background**
- A URL pública será gerada automaticamente
- Use **Ctrl+C** para parar o servidor se necessário

````python
# 🚀 ETAPA 3.1: Criar servidor Flask integrado para AEON
print("🔧 Configurando servidor Flask para AEON...")

from flask import Flask, render_template_string, jsonify, request
from flask_ngrok import run_with_ngrok
import threading
import time
import json
import os

# Criar aplicação Flask
app = Flask(__name__)
run_with_ngrok(app)  # Configurar ngrok

# Template HTML da interface AEON
AEON_TEMPLATE = """
<!DOCTYPE html>
<html lang="pt-BR">
<head>
    <meta charset="UTF-8">
    <meta name="viewport" content="width=device-width, initial-scale=1.0">
    <title>🧬 AEON - Digital Twin MVP</title>
    <style>
        body { font-family: 'Segoe UI', Tahoma, Geneva, Verdana, sans-serif; 
               background: linear-gradient(135deg, #1e3c72, #2a5298); 
               color: white; margin: 0; padding: 20px; }
        .container { max-width: 1200px; margin: 0 auto; }
        .header { text-align: center; margin-bottom: 30px; }
        .module { background: rgba(255,255,255,0.1); border-radius: 15px; 
                  padding: 20px; margin: 15px 0; backdrop-filter: blur(10px); }
        .status { display: inline-block; padding: 5px 15px; border-radius: 20px; 
                  font-size: 12px; margin: 5px; }
        .active { background: #28a745; }
        .inactive { background: #6c757d; }
        button { background: #007bff; color: white; border: none; 
                 padding: 10px 20px; border-radius: 5px; cursor: pointer; margin: 5px; }
        button:hover { background: #0056b3; }
        .api-result { background: rgba(0,0,0,0.3); padding: 15px; 
                      border-radius: 10px; margin: 10px 0; font-family: monospace; }
    </style>
</head>
<body>
    <div class="container">
        <div class="header">
            <h1>🧬 AEON - Digital Twin MVP</h1>
            <p>Advanced Entropy Optimization Network - Sistema Completo</p>
        </div>
        
        <div class="module">
            <h2>🔬 Módulos Ativos</h2>
            <span class="status active">🧬 Análise de Entropia</span>
            <span class="status active">🧠 V.E.R.N.A. AI</span>
            <span class="status active">🌌 Cosmologia NMD</span>
            <span class="status active">⚡ Digital Twin UHE</span>
            <span class="status active">🌐 Interface Web</span>
        </div>
        
        <div class="module">
            <h2>🚀 Testes de API</h2>
            <button onclick="testAPI('/api/status')">Status do Sistema</button>
            <button onclick="testAPI('/api/entropia')">Análise Entropia</button>
            <button onclick="testAPI('/api/verna')">V.E.R.N.A. Report</button>
            <button onclick="testAPI('/api/cosmologia')">Cosmologia NMD</button>
            <button onclick="testAPI('/api/simulation')">Simulação Completa</button>
            <div id="api-results" class="api-result" style="display:none;">
                <h3>📊 Resultado da API:</h3>
                <pre id="result-content"></pre>
            </div>
        </div>
        
        <div class="module">
            <h2>📈 Status do Digital Twin</h2>
            <p><strong>🕒 Uptime:</strong> <span id="uptime">Calculando...</span></p>
            <p><strong>🔄 Última Atualização:</strong> <span id="last-update">Agora</span></p>
            <p><strong>📊 Simulações Executadas:</strong> <span id="sim-count">0</span></p>
        </div>
    </div>
    
    <script>
        let startTime = new Date();
        let simCount = 0;
        
        function updateUptime() {
            const now = new Date();
            const diff = now - startTime;
            const seconds = Math.floor(diff / 1000);
            const minutes = Math.floor(seconds / 60);
            const hours = Math.floor(minutes / 60);
            document.getElementById('uptime').textContent = 
                hours + 'h ' + (minutes % 60) + 'm ' + (seconds % 60) + 's';
        }
        
        function testAPI(endpoint) {
            fetch(endpoint)
                .then(response => response.json())
                .then(data => {
                    document.getElementById('api-results').style.display = 'block';
                    document.getElementById('result-content').textContent = 
                        JSON.stringify(data, null, 2);
                    if (endpoint === '/api/simulation') simCount++;
                    document.getElementById('sim-count').textContent = simCount;
                    document.getElementById('last-update').textContent = new Date().toLocaleString();
                })
                .catch(error => {
                    document.getElementById('api-results').style.display = 'block';
                    document.getElementById('result-content').textContent = 
                        'Erro: ' + error.message;
                });
        }
        
        setInterval(updateUptime, 1000);
    </script>
</body>
</html>
"""

print("✅ Template HTML configurado!")
````

````python
# 🚀 ETAPA 3.2: Configurar rotas da API AEON
print("🔗 Configurando rotas da API...")

# Rota principal - Interface Web
@app.route('/')
def home():
    return render_template_string(AEON_TEMPLATE)

# API: Status do Sistema
@app.route('/api/status')
def api_status():
    return jsonify({
        "system": "AEON Digital Twin MVP",
        "version": "1.0.0",
        "status": "online",
        "modules": {
            "entropia": "active",
            "verna": "active", 
            "cosmologia": "active",
            "digital_twin": "active",
            "web_interface": "active"
        },
        "timestamp": time.strftime("%Y-%m-%d %H:%M:%S")
    })

# API: Análise de Entropia
@app.route('/api/entropia')
def api_entropia():
    import random
    return jsonify({
        "module": "Análise de Entropia",
        "entropia_atual": round(random.uniform(0.7, 0.95), 3),
        "tendencia": "crescente",
        "predicao_24h": round(random.uniform(0.8, 1.0), 3),
        "confiabilidade": "alta",
        "parametros": {
            "temperatura": round(random.uniform(20, 30), 1),
            "pressao": round(random.uniform(1000, 1100), 2),
            "umidade": round(random.uniform(40, 70), 1)
        }
    })

# API: V.E.R.N.A. Report
@app.route('/api/verna')
def api_verna():
    return jsonify({
        "module": "V.E.R.N.A. - Virtual Entity for Rational Network Analysis",
        "analysis_type": "network_optimization",
        "recommendations": [
            "Otimizar conexões P2P",
            "Reduzir latência em 15%",
            "Implementar cache distribuído"
        ],
        "network_health": "excelente",
        "efficiency_score": 92.5,
        "active_nodes": 847,
        "data_processed": "15.7 TB"
    })

print("✅ Rotas da API configuradas!")
````

````python
# 🚀 ETAPA 3.3: Configurar rotas adicionais da API
print("🌌 Adicionando rotas de Cosmologia e Simulação...")

# API: Cosmologia NMD
@app.route('/api/cosmologia')
def api_cosmologia():
    import random
    return jsonify({
        "module": "Cosmologia NMD - Non-linear Multidimensional Dynamics",
        "universe_model": "Lambda-CDM",
        "dark_matter_density": round(random.uniform(0.25, 0.30), 3),
        "dark_energy_density": round(random.uniform(0.68, 0.72), 3),
        "hubble_constant": round(random.uniform(67, 73), 1),
        "age_universe": "13.8 bilhões de anos",
        "predictions": {
            "expansion_rate": "acelerando",
            "big_rip_probability": "baixa",
            "heat_death_timeline": "10^100 anos"
        }
    })

# API: Simulação Completa
@app.route('/api/simulation')
def api_simulation():
    import random
    return jsonify({
        "simulation_id": f"SIM_{int(time.time())}",
        "type": "digital_twin_complete",
        "duration": "5.2 segundos",
        "modules_executed": ["entropia", "verna", "cosmologia"],
        "results": {
            "system_efficiency": round(random.uniform(85, 95), 1),
            "optimization_gain": round(random.uniform(10, 25), 1),
            "energy_saved": f"{round(random.uniform(100, 500), 1)} kWh",
            "cost_reduction": f"${round(random.uniform(1000, 5000), 2)}"
        },
        "recommendations": [
            "Aumentar paralelização em 20%",
            "Implementar cache L2",
            "Otimizar algoritmos de entropia"
        ],
        "next_simulation": "em 15 minutos"
    })

print("✅ Todas as rotas da API configuradas!")
print("🎯 Pronto para iniciar servidor com ngrok!")
````

````python
# 🚀 ETAPA 3.4: INICIAR SERVIDOR AEON COM NGROK
print("🌐 Iniciando servidor AEON com acesso público...")
print("⚡ O servidor ficará rodando e gerará uma URL pública!")
print("🔗 A URL aparecerá logo abaixo...")
print("")
print("🚨 IMPORTANTE: Após executar esta célula:")
print("   1. Uma URL pública será gerada (https://xxxx.ngrok.io)")
print("   2. Clique na URL para acessar a interface AEON")
print("   3. Teste todas as APIs através da interface")
print("   4. Use Ctrl+C para parar o servidor quando necessário")
print("")
print("🎯 Executando servidor...")

# Iniciar servidor Flask com ngrok
app.run()
````

# 🔗 Passo 4: Extrair URL e Testar APIs

Após executar a célula anterior, o **ngrok** gerará uma URL pública para acessar o AEON.

1. **Copie a URL gerada** (formato: `https://xxxx.ngrok.io`)
2. **Abra em nova aba** para acessar a interface completa

## 🧪 Testes Automatizados:

Execute as células abaixo para testar todas as APIs automaticamente.

````python
# 🚀 ETAPA 4.1: Obter URL pública do ngrok
print("🔍 Buscando URL pública do ngrok...")

import requests
import json

try:
    # Obter informações do ngrok
    response = requests.get("http://localhost:4040/api/tunnels")
    if response.status_code == 200:
        tunnels = response.json()['tunnels']
        if tunnels:
            public_url = tunnels[0]['public_url']
            print(f"✅ URL PÚBLICA ENCONTRADA:")
            print(f"🌐 {public_url}")
            print(f"")
            print(f"🎯 ACESSE AGORA: {public_url}")
            print(f"")
            print(f"📋 URLs das APIs:")
            print(f"   • Status: {public_url}/api/status")
            print(f"   • Entropia: {public_url}/api/entropia") 
            print(f"   • V.E.R.N.A.: {public_url}/api/verna")
            print(f"   • Cosmologia: {public_url}/api/cosmologia")
            print(f"   • Simulação: {public_url}/api/simulation")
            
            # Salvar URL para próximas células
            globals()['AEON_URL'] = public_url
        else:
            print("❌ Nenhum túnel encontrado. Execute a célula do servidor primeiro!")
    else:
        print("❌ Erro ao conectar com ngrok. Verifique se o servidor está rodando.")
except Exception as e:
    print(f"❌ Erro: {e}")
    print("💡 Dica: Execute a célula do servidor primeiro!")
````

````python
# 🚀 ETAPA 4.2: Testes automáticos das APIs AEON
print("🧪 Executando testes automáticos das APIs...")

import requests
import json
import time

def test_api_endpoint(url, name):
    """Função para testar endpoint da API"""
    try:
        print(f"\n🔬 Testando {name}...")
        response = requests.get(url, timeout=10)
        
        if response.status_code == 200:
            data = response.json()
            print(f"✅ {name} - Status: OK")
            print(f"📊 Dados recebidos: {len(str(data))} caracteres")
            
            # Mostrar dados principais
            if 'system' in data:
                print(f"   Sistema: {data['system']}")
            if 'module' in data:
                print(f"   Módulo: {data['module']}")
            if 'simulation_id' in data:
                print(f"   ID Simulação: {data['simulation_id']}")
                
            return True
        else:
            print(f"❌ {name} - Erro HTTP: {response.status_code}")
            return False
            
    except Exception as e:
        print(f"❌ {name} - Erro: {str(e)}")
        return False

# Verificar se temos a URL
if 'AEON_URL' in globals():
    base_url = AEON_URL
    print(f"🎯 Testando APIs em: {base_url}")
    
    # Lista de endpoints para testar
    endpoints = [
        ("/api/status", "Status do Sistema"),
        ("/api/entropia", "Análise de Entropia"),
        ("/api/verna", "V.E.R.N.A. Report"),
        ("/api/cosmologia", "Cosmologia NMD"),
        ("/api/simulation", "Simulação Completa")
    ]
    
    successful_tests = 0
    total_tests = len(endpoints)
    
    for endpoint, name in endpoints:
        if test_api_endpoint(base_url + endpoint, name):
            successful_tests += 1
        time.sleep(1)  # Pausa entre testes
    
    print(f"\n🏆 RESULTADO DOS TESTES:")
    print(f"✅ Sucessos: {successful_tests}/{total_tests}")
    print(f"📊 Taxa de sucesso: {(successful_tests/total_tests)*100:.1f}%")
    
    if successful_tests == total_tests:
        print(f"🎉 TODOS OS TESTES PASSARAM! AEON está funcionando perfeitamente!")
    else:
        print(f"⚠️  Alguns testes falharam. Verifique se o servidor está rodando.")
        
else:
    print("❌ URL não encontrada. Execute a célula anterior primeiro!")
````
